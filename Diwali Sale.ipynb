```python
import numpy as np
import pandas as pd 
import plotly.express as px
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Sale_Data = pd.read_csv(r"C:/Users/lenovo/Downloads/Diwali_Sales_Data.csv", encoding='ISO-8859-1')
Sale_Data.head()
```

```python
Sale_Data.shape
```

```python
Sale_Data.info()
```

```python
Sale_Data.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
pd.isnull(Sale_Data).sum()
```

```python
Sale_Data.dropna(inplace=True)
```

```python
#change the data type 
Sale_Data["Amount"]=Sale_Data["Amount"].astype('int')
```

```python
Sale_Data['Amount'].dtypes
```

```python
Sale_Data.columns
```

```python
# Renme the column 
Sale_Data.rename(columns={'Marital_Status': "Married/Unmarried"})
```

```python
Sale_Data.describe()
```

```python
Sale_Data[['Age','Orders','Amount']].describe()
```

```python
#exploratory data analysis
Sale_Data.columns
```

```python
ax=sns.countplot(x='Gender',data=Sale_Data)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
Sale_gen=Sale_Data.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.barplot(x='Gender',y='Amount',data=Sale_gen)
```

```python
#Age
ax=sns.countplot(x='Age Group',hue="Gender" ,data=Sale_Data)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
Sale_age=Sale_Data.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.barplot(x='Age Group',y='Amount',data=Sale_age)
```

from the above graph we can see that most of the buyer are of the age 26-35 yrs female 

```python
#state 

Sale_State=Sale_Data.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='State',y='Orders',data=Sale_State)
```

```python
Sale_State1=Sale_Data.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='State',y='Amount',data=Sale_State1)
```

The top 3 state which are having having the highest purchase is Uttar Pradesh, Maharashtra,Karnataka

```python
#Marital Status
ax=sns.countplot(x='Marital_Status' ,data=Sale_Data)
sns.set(rc={'figure.figsize':(6,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
Sale_State1=Sale_Data.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)

sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(x='Marital_Status',y='Amount',data=Sale_State1,hue='Gender')
```

From the above graph we can see that most buyers are the married women and have higher parchasing power 

```python
#occupation
ax=sns.countplot(x='Occupation' ,data=Sale_Data)
sns.set(rc={'figure.figsize':(20,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
Sale_State1=Sale_Data.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Occupation',y='Amount',data=Sale_State1)
```

From the above graph we can see that most of the buyers are from IT sector ,Healthcare,Aviation sector

```python
#Product Category
ax=sns.countplot(x='Product_Category' ,data=Sale_Data)
sns.set(rc={'figure.figsize':(70,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
Sale_State1=Sale_Data.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(x='Product_Category',y='Amount',data=Sale_State1)
```

From the above graph we can see that most of the product sold are in food , clothing and apparel and electronics 

#conclusion

Married women age group between 26-35 yrs from UP, Maharastha and Karnataka working in IT ,healthcare and Aviation are mostly buy products like foods ,clothing and electronic category
