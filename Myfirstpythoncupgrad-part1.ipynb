```python
#nested loop
rows = 6

columns = 6



i = 1

while i <= rows:

    j = 1

    while j <= columns:

        print(str(i * j), end = ' ')

        j += 1

    print()

    i += 1





rows = 6

columns = 6



for i in range(1, rows + 1):

    for j in range(1, columns + 1):

        print(str(i * j), end = ' ')

    print()
```

```python
def sum_of_cubes(start, end, step = 1):
    total = 0
    for i in range(start, end, step):
        total += i ** 3
    return total

result1 = sum_of_cubes(1, 4)
result2 = sum_of_cubes(2, 6, 2)

print(result1, result2)
```

##Lambda expression

```python
#lamda is the alternative to function in python
#lamda is used to create 1 line functon
#function_name=lambda <input_variable> : <output_variable>
getsquare= lambda  num :  num *num
result= getsquare(5)
print(result)
```

```python
poundtokg=lambda lbs :lbs*0.45
result=poundtokg(110)
print(result)
```

```python
isEvenorodd=lambda number :'even' if number % 2 == 0 else 'odd'
result=isEvenorodd(133)
print(result)
```

```python
isEven= lambda num : num**2 if num % 2 == 0 else num**3
result=isEven(4)
print(result)
```
