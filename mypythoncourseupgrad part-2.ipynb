```python
input_str = "I love programming in python"

count=0    # initializing count variable



l = ['a', 'e', 'i', 'o', 'u']

m = ['y']



# Converting the string to lowercase

for i in input_str.lower():

  if i in l:

    count = count+1

  if i in m:

    count = count-1

print(count)
```

#Dictionary
# A dictionary is represented using {KEY : Value}
#1. A Dictionary is Represented using  { KEY : Value }
2. You cannot Repeat Key in a Dictionary.
3. Dictionary is Mutable Data Structure.
4. Dictionary is Ordered Data Structure.
5. You can use Only Immutable DataTypes as Keys , Mutable DataTypes as Keys aren't Allowed.

  eg. { 'a': 'apple' } here 'a' is string and string is immutable hence valid Key ✅


  { ['a'] : 'apple' }  here ['a'] is list and list is Mutable hence Invalid KEY. ❌
