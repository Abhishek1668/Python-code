Performance test 
objective to add 1 million number using python vs numpy,one taking less time to complete opertion will be the winner 

```python
import time
import numpy as np

SIZE = 1000000

A1 = range( SIZE )
A2 = range( SIZE )

L1 = np.arange( SIZE )
L2 = np.arange( SIZE )

# Python Test
start = time.time()
result= [x+y for x,y in zip(A1,A2)]
stop = time.time()

print('🐍 Python took ', (stop-start) * 1000 , ' ms')

# Numpy Test

start = time.time()
result = L1 + L2
stop = time.time()

print('👨‍🔬 Numpy Took ' , (stop-start) * 1000,' ms')
```

Suppose a Fruit Seller Sold 20 Mangoes and 10 Oranges in one Day for a total of 350 . the next day he sold 17 mangoes and 22 oranges for 500. if the price of fruit remains unchanged on both days , what was the price of one mango 🥭 and one orange 🍊

```python
#20x + 10y =350
#17x + 22y=500
```

```python
coeff=np.array([[20,10],[17,22]])
```

```python
const=np.array([350,500])
```

```python
result=np.linalg.solve(coeff,const)
print(result)
```
